calcul du mode et remplacement des valeurs manquanytes par le mode

```python
mode_htn=data1["htn"].mode()[0]
mode_dm=data1["dm"].mode()[0]
mode_cad=data1["cad"].mode()[0]
mode_bgr=data1["bgr"].median()
```

```python
data1["htn"]=data1["htn"].fillna(mode_htn)
data1["dm"]=data1["dm"].fillna(mode_dm)
data1["cad"]=data1["cad"].fillna(mode_cad)
data1["bgr"]=data1["bgr"].fillna(mode_bgr)
```

```python
data1.info()
```

remplacement des valeurs manquantes par la mediane

```python
data1["hemo"]=data1["hemo"].fillna(data1["hemo"].median())
data1["sc"]=data1["sc"].fillna(data1["sc"].median())
data1["bp"]=data1["bp"].fillna(data1["bp"].median())
```

```python
data1.info()
```

```python
data1["age"]=data1["age"].fillna(data1["age"].median())
data1["bu"]=data1["bu"].fillna(data1["bu"].median())
data1["al"]=data1["al"].fillna(data1["al"].median())
```

### Code des Boxplots pour toutes les variables

```python
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Sélectionner uniquement les variables numériques de data1 (exclure 'classification')
numeric_cols = data1.select_dtypes(include=['float64', 'int64']).columns.tolist()

# Déterminer la disposition de la grille de graphiques (3 colonnes par ligne)
n_cols = 3
n_rows = math.ceil(len(numeric_cols) / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows))
axes = axes.flatten()  # Aplatir la grille pour faciliter l'itération

for i, col in enumerate(numeric_cols):
    sns.boxplot(ax=axes[i], x='classification', y=col, data=data1, hue='classification', palette='Set2', legend=False)
    axes[i].set_title(f'Distribution de {col}')
    axes[i].set_xlabel('Classification')
    axes[i].set_ylabel(col)

# Masquer les graphiques vides si le nombre de variables n'est pas un multiple de 3
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
