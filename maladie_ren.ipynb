calcul du mode et remplacement des valeurs manquanytes par le mode

```python
mode_htn=data1["htn"].mode()[0]
mode_dm=data1["dm"].mode()[0]
mode_cad=data1["cad"].mode()[0]
mode_bgr=data1["bgr"].median()
```

```python
data1["htn"]=data1["htn"].fillna(mode_htn)
data1["dm"]=data1["dm"].fillna(mode_dm)
data1["cad"]=data1["cad"].fillna(mode_cad)
data1["bgr"]=data1["bgr"].fillna(mode_bgr)
```

```python
data1["hemo"]=data1["hemo"].fillna(data1["hemo"].median())
data1["sc"]=data1["sc"].fillna(data1["sc"].median())
data1["bp"]=data1["bp"].fillna(data1["bp"].median())
```

### Code des Boxplots pour toutes les variables

```python
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Sélectionner uniquement les variables numériques de data1 (exclure 'classification')
numeric_cols = data1.select_dtypes(include=['float64', 'int64']).columns.tolist()

# Déterminer la disposition de la grille de graphiques (3 colonnes par ligne)
n_cols = 3
n_rows = math.ceil(len(numeric_cols) / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows))
axes = axes.flatten()  # Aplatir la grille pour faciliter l'itération

for i, col in enumerate(numeric_cols):
    sns.boxplot(ax=axes[i], x='classification', y=col, data=data1, hue='classification', palette='Set2', legend=False)
    axes[i].set_title(f'Distribution de {col}')
    axes[i].set_xlabel('Classification')
    axes[i].set_ylabel(col)

# Masquer les graphiques vides si le nombre de variables n'est pas un multiple de 3
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

### Visualisation des variables catégorielles (Diagrammes à barres)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Liste des variables catégorielles d'intérêt d'origine
cat_vars = ['htn', 'dm', 'cad']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes = axes.flatten()

for i, col in enumerate(cat_vars):
    sns.countplot(ax=axes[i], x=col, hue='classification', data=data1, palette='Set2')
    axes[i].set_title(f'Distribution de {col} par Classification')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Nombre')

plt.tight_layout()
plt.show()
```

### Nettoyage et affichage propre de la distribution des variables catégorielles

Nous allons supprimer les espaces blancs et les tabulations pour regrouper correctement les modalités identiques.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Nettoyer les colonnes catégorielles pour enlever les espaces et tabulations indésirables
for col in ['htn', 'dm', 'cad', 'classification']:
    data1[col] = data1[col].astype(str).str.strip()

# Recréer les graphiques avec les données nettoyées
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes = axes.flatten()

for i, col in enumerate(['htn', 'dm', 'cad']):
    sns.countplot(ax=axes[i], x=col, hue='classification', data=data1, palette='Set2')
    axes[i].set_title(f'Distribution de {col} par Classification (Nettoyé)')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Nombre')

plt.tight_layout()
plt.show()
```

### Visualisation des pourcentages par Diagrammes en Camembert (Pie Charts)

```python
import matplotlib.pyplot as plt


cat_vars = ['htn', 'dm', 'cad', 'classification']

fig, axes = plt.subplots(2, 2, figsize=(14, 12))
axes = axes.flatten()

for i, col in enumerate(cat_vars):

    counts = data1[col].value_counts()

    # Dessiner le diagramme en camembert
    axes[i].pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Set3.colors)
    axes[i].set_title(f'Répartition de {col}')

plt.tight_layout()
plt.show()
```
